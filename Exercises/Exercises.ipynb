# PageRank Exercises

Definition of the power method used to solve the following exercises

```python
import numpy as np

def power_method(A, tol=1e-12, iterations=1000):
    xk = np.ones((A.shape[0], 1))
    xk /= np.sum(xk) 
    for i in range(iterations):
        xk_old = xk.copy()
        xk = (A @ xk)
        xk /= np.sum(xk)
        residual = np.linalg.norm(xk - xk_old, 1)
        if residual < tol:
            break            
    return xk
```

# Exercise 1

```python
A = np.array([
    [0, 0, 1/2, 1/2, 0],
    [1/3, 0, 0, 0, 0],
    [1/3, 1/2, 0, 1/2, 1],
    [1/3, 1/2, 0, 0, 0],
    [0, 0, 1/2, 0, 0]]) 

#-------------------------
# Computation of rankings 
#-------------------------
eigenvalues, eigenvectors = np.linalg.eig(A) 
rk = eigenvectors[:, np.argmax(eigenvalues.real)].real 
rk /= np.sum(rk) 
rk_pm = power_method(A)

print("Ranking with linalg library:", np.round(rk, 5))

print("\nRanking computed using our self implemented power method: ")
print(np.round(rk_pm.flatten(), 5))
```

# Exercise 16
Consider the link matrix A (see code below).
Show that M = (1− m)A + mS (all Sij = 1/3) is not diagonalizable for 0 ≤ m < 1.

Given a generic matrix $A \in \mathbb{R}^{n \times n}$, A is not diagonalizable iff the geometric multiplicity of at least one of its eigenvalues is less than its algebraic multiplicity. 
- Algebraic multiplicity of $\lambda_i$ is the integer $\alpha$ such that $p(\lambda) = (\lambda - \lambda_i)^{\alpha} q(\lambda)$ with $q(\lambda) \neq 0$
- Geometric multiplicity of $\lambda_i$ is the number of linearly independent eigenvectors corresponding to $\lambda_{i}$

Since this is a necessary condition to guarantee diagonalizability of any matrix, if we want to show that $M = (1-m)A + mS$ is non diagonalizable it's sufficient to find one eigenvalue of M whose algebraic and geometric multiplicity don't coincide.

$$
M = (1-m) \begin{bmatrix}
0 & \frac{1}{2} & \frac{1}{2} \\
0 & 0 & \frac{1}{2} \\
1 & \frac{1}{2} & 0
\end{bmatrix}
+ m \begin{bmatrix}
\frac{1}{3} & \frac{1}{3} & \frac{1}{3} \\
\frac{1}{3} & \frac{1}{3} & \frac{1}{3} \\
\frac{1}{3} & \frac{1}{3} & \frac{1}{3}
\end{bmatrix}
= \begin{bmatrix}
\frac{m}{3} & \frac{3-m}{6} & \frac{3-m}{6} \\
\frac{m}{3} & \frac{m}{3} & \frac{3-m}{6} \\
1 - \frac{2m}{3} & \frac{3-m}{6} & \frac{m}{3}
\end{bmatrix}
$$





```python
from sympy import symbols, Matrix

m = symbols('m')
M = Matrix([
    [m/3, (3-m)/6, (3-m)/6],
    [m/3, m/3, (3-m)/6],
    [1-2*m/3, (3-m)/6, m/3]
])

lambdas = M.eigenvals()
v = M.eigenvects()
```

$\lambda_1 = 1$ has algebraic multiplicity 1

$\lambda_2 = (\frac{m}{2}-\frac{1}{2})$ has algebraic multiplicity 2


Both eigenvectors appear just once, meaning that both their respective eigenvalues have **geometric multiplicity equal to one**. Showing that the second eigenvalue has different algebraic and gemoetric multiplicities proves that, for any value of $m \in [0, 1)$ the matrix M is not diagonalizable.
